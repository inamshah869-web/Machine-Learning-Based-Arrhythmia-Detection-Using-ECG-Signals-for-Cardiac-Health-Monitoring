```python
# 1. Install necessary external packages first
%pip install pytorch-tabnet
%pip install imbalanced-learn

#---Essential Code for Report Appendix--

# 2. Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
import torch
from pytorch_tabnet.tab_model import TabNetClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
import lightgbm as lgb
from sklearn.tree import DecisionTreeClassifier
import warnings

# Suppress potential warnings for cleaner output in the report
warnings.filterwarnings('ignore')


#---1. Data Loading and Preprocessing--

# Load the dataset
try:
    df = pd.read_csv('/content/arrhythmia.csv')
except FileNotFoundError:
    print("Error: arrhythmia.csv not found. Please ensure the file is in the correct directory.")
    exit()

# Handle Missing Values and Impute
df_cleaned = df.replace('?', np.nan)
X = df_cleaned.drop('class', axis=1).astype(float)
y = df_cleaned['class'].astype(int)

imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
X_imputed = imputer.fit_transform(X)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# Apply PCA before SMOTE (Optimized order)
pca = PCA(n_components=100, random_state=42)
X_pca_before_smote = pca.fit_transform(X_scaled)

# Apply SMOTE on the PCA-transformed data
majority_class_count = y.value_counts().max()
sampling_strategy = {cls: majority_class_count for cls in y.unique()}
min_samples_in_oversampled_classes = min(
    y.value_counts()[list(sampling_strategy.keys())]
)
smote_k = max(min_samples_in_oversampled_classes - 1, 1)

smote = SMOTE(
    random_state=42,
    sampling_strategy=sampling_strategy,
    k_neighbors=smote_k
)

X_resampled_after_pca, y_resampled_after_pca = smote.fit_resample(
    X_pca_before_smote, y
)

# Split the resampled data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled_after_pca,
    y_resampled_after_pca,
    test_size=0.2,
    random_state=42,
    stratify=y_resampled_after_pca
)

# Create indexed labels for Keras/TabNet models (0-indexed)
unique_classes_train = sorted(np.unique(y_train))
class_map = {cls: i for i, cls in enumerate(unique_classes_train)}
y_train_indexed = y_train.map(class_map).values
y_test_indexed = y_test.map(class_map).values

# Create mapping from indexed labels back to original labels for reporting
indexed_to_original_mapping = {i: cls for cls, i in class_map.items()}

# Reshape data for CNN and LSTM (assuming PCA components are treated as timesteps with 1 feature)
X_train_reshaped_cnn = X_train.reshape(
    (X_train.shape[0], X_train.shape[1], 1)
)
X_test_reshaped_cnn = X_test.reshape(
    (X_test.shape[0], X_test.shape[1], 1)
)


X_train_reshaped_lstm = X_train.reshape(
    (X_train.shape[0], X_train.shape[1], 1)
)
X_test_reshaped_lstm = X_test.reshape(
    (X_test.shape[0], X_test.shape[1], 1)
)


#---2. Helper Functions for Model Training and Evaluation--

def get_evaluation_metrics(y_true, y_pred, model_name, labels_type='original'):
    """
    Calculates evaluation metrics (Accuracy, Macro Avg F1, Weighted Avg F1).

    Args:
        y_true: True labels (original or indexed).
        y_pred: Predicted labels (original or indexed).
        model_name: Name of the model for reporting.
        labels_type: 'original' or 'indexed' to specify label type for report.

    Returns:
        dict: A dictionary containing the model name and key metrics.
    """
    report = classification_report(
        y_true,
        y_pred,
        output_dict=True,
        zero_division=0
    )

    accuracy = report.get('accuracy', 0.0)
    macro_f1 = report.get('macro avg', {}).get('f1-score', 0.0)
    weighted_f1 = report.get('weighted avg', {}).get('f1-score', 0.0)

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Macro Avg F1': macro_f1,
        'Weighted Avg F1': weighted_f1,
        'y_true': y_true,
        'y_pred': y_pred
    }


def plot_confusion_matrix(y_true, y_pred, title, labels):
    """
    Generates and displays a confusion matrix.

    Args:
        y_true: True labels.
        y_pred: Predicted labels.
        title: Plot title.
        labels: List of class labels for axis ticks (original or indexed).
    """
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        xticklabels=labels,
        yticklabels=labels
    )
    plt.title(title)
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    # Save figure before displaying
    filename = f"{title.replace(' ', '_').replace(':', '')}.png"
    plt.savefig(filename, bbox_inches='tight')
    plt.show()


def train_evaluate_sklearn_model(
    model,
    X_train,
    y_train,
    X_test,
    y_test,
    model_name
):
    """
    Trains and evaluates a scikit-learn classification model.
    """
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    metrics = get_evaluation_metrics(
        y_test,
        y_pred,
        model_name,
        labels_type='original'
    )
    return metrics


def build_keras_mlp_model(
    input_shape,
    num_classes,
    dense_layers=(128, 64),
    dropout_rate=0.3,
    optimizer='adam',
    learning_rate=0.001
):
    """Builds a Keras MLP model."""
    model = keras.Sequential()
    model.add(keras.layers.InputLayer(input_shape=input_shape))

    for layer_size in dense_layers:
        model.add(keras.layers.Dense(layer_size, activation='relu'))
        model.add(keras.layers.Dropout(dropout_rate))

    model.add(keras.layers.Dense(num_classes, activation='softmax'))

    if optimizer == 'adam':
        opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer == 'rmsprop':
        opt = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
    else:
        opt = optimizer

    model.compile(
        optimizer=opt,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


def build_keras_cnn_model(
    input_shape,
    num_classes,
    filters_1=32,
    kernel_size_1=3,
    pool_size_1=2,
    filters_2=64,
    kernel_size_2=3,
    pool_size_2=2,
    dense_units=100,
    dropout_rate=0.5,
    optimizer='adam',
    learning_rate=0.001
):
    """Builds a Keras 1D CNN model."""
    model = keras.Sequential([
        keras.layers.InputLayer(input_shape=input_shape),
        keras.layers.Conv1D(
            filters=filters_1,
            kernel_size=kernel_size_1,
            activation='relu'
        ),
        keras.layers.MaxPooling1D(pool_size=pool_size_1),
        keras.layers.Conv1D(
            filters=filters_2,
            kernel_size=kernel_size_2,
            activation='relu'
        ),
        keras.layers.MaxPooling1D(pool_size=pool_size_2),
        keras.layers.Flatten(),
        keras.layers.Dense(dense_units, activation='relu'),
        keras.layers.Dropout(dropout_rate),
        keras.layers.Dense(num_classes, activation='softmax')
    ])

    if optimizer == 'adam':
        opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer == 'rmsprop':
        opt = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
    else:
        opt = optimizer

    model.compile(
        optimizer=opt,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


def build_keras_lstm_model(
    input_shape,
    num_classes,
    lstm_units=100,
    dense_units=50,
    dropout_rate=0.5,
    optimizer='adam',
    learning_rate=0.001
):
    """Builds a Keras LSTM model."""
    model = keras.Sequential([
        keras.layers.InputLayer(input_shape=input_shape),
        keras.layers.LSTM(
            units=lstm_units,
            return_sequences=False
        ),
        keras.layers.Dense(dense_units, activation='relu'),
        keras.layers.Dropout(dropout_rate),
        keras.layers.Dense(num_classes, activation='softmax')
    ])

    if optimizer == 'adam':
        opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer == 'rmsprop':
        opt = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
    else:
        opt = optimizer

    model.compile(
        optimizer=opt,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


def train_evaluate_keras_model(
    model,
    X_train,
    y_train_indexed,
    X_test,
    y_test_indexed,
    model_name="Keras Model",
    epochs=50,
    batch_size=32
):
    """
    Trains and evaluates a Keras classification model.
    """
    model.fit(
        X_train,
        y_train_indexed,
        epochs=epochs,
        batch_size=batch_size,
        verbose=0
    )

    y_pred_prob = model.predict(X_test)
    y_pred = tf.argmax(y_pred_prob, axis=1).numpy()

    metrics = get_evaluation_metrics(
        y_test_indexed,
        y_pred,
        model_name,
        labels_type='indexed'
    )

    return metrics


def train_evaluate_tabnet_model(
    model,
    X_train_np,
    y_train_np,
    X_test_np,
    y_test_np,
    model_name="TabNet Model"
):
    """
    Trains and evaluates a TabNet classification model.
    """
    model.fit(
        X_train=X_train_np,
        y_train=y_train_np,
        eval_set=[(X_test_np, y_test_np)],
        eval_metric=['accuracy'],
        max_epochs=100,
        batch_size=1024,
        virtual_batch_size=128
    )

    y_pred = model.predict(X_test_np)

    metrics = get_evaluation_metrics(
        y_test_np,
        y_pred,
        model_name,
        labels_type='indexed'
    )

    return metrics


#---3. Model Training and Evaluation--

all_model_performance = []


#---Train and Evaluate Scikit-learn Models--

# Tuned SVM (using best parameters found previously)
tuned_svm_model = SVC(
    C=100,
    kernel='rbf',
    probability=True,
    random_state=42
)

svm_metrics = train_evaluate_sklearn_model(
    tuned_svm_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "Tuned SVM"
)

all_model_performance.append(svm_metrics)


# RandomForestClassifier (using default parameters)
rf_model = RandomForestClassifier(random_state=42)

rf_metrics = train_evaluate_sklearn_model(
    rf_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "RandomForestClassifier"
)

all_model_performance.append(rf_metrics)


# LogisticRegression (using default parameters, increased max_iter)
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_metrics = train_evaluate_sklearn_model(
    lr_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "LogisticRegression"
)

all_model_performance.append(lr_metrics)


# KNeighborsClassifier (using default parameters)
knn_model = KNeighborsClassifier()

knn_metrics = train_evaluate_sklearn_model(
    knn_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "KNeighborsClassifier"
)

all_model_performance.append(knn_metrics)


# LightGBM (using default parameters)
lgb_model = lgb.LGBMClassifier(random_state=42)

lgb_metrics = train_evaluate_sklearn_model(
    lgb_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "LightGBM"
)

all_model_performance.append(lgb_metrics)


# AdaBoostClassifier (using default parameters)
ada_model = AdaBoostClassifier(random_state=42)

ada_metrics = train_evaluate_sklearn_model(
    ada_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "AdaBoostClassifier"
)

all_model_performance.append(ada_metrics)


# DecisionTreeClassifier (using default parameters)
dt_model = DecisionTreeClassifier(random_state=42)

dt_metrics = train_evaluate_sklearn_model(
    dt_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "DecisionTreeClassifier"
)

all_model_performance.append(dt_metrics)


#---Train and Evaluate Keras/TensorFlow Models--

# Tuned MLP (using parameters from Keras Tuner or manual fallback)
mlp_params = {
    'dense_layers': (192,),
    'dropout_rate': 0.4,
    'optimizer': 'rmsprop',
    'learning_rate': 0.002177994311825884,
    'epochs': 50,
    'batch_size': 32
}

mlp_model = build_keras_mlp_model(
    input_shape=(X_train.shape[1],),
    num_classes=len(np.unique(y_train_indexed)),
    dense_layers=mlp_params['dense_layers'],
    dropout_rate=mlp_params['dropout_rate'],
    optimizer=mlp_params['optimizer'],
    learning_rate=mlp_params['learning_rate']
)

mlp_metrics = train_evaluate_keras_model(
    mlp_model,
    X_train,
    y_train_indexed,
    X_test,
    y_test_indexed,
    epochs=mlp_params['epochs'],
    batch_size=mlp_params['batch_size'],
    model_name="Tuned MLP"
)

all_model_performance.append(mlp_metrics)


# TabNet (using default parameters)
if torch.cuda.is_available():
    device = 'cuda'
else:
    device = 'cpu'

tabnet_model = TabNetClassifier(
    n_d=64,
    n_a=64,
    optimizer_fn=torch.optim.Adam,
    optimizer_params=dict(lr=2e-2),
    scheduler_params={"step_size": 50, "gamma": 0.9},
    scheduler_fn=torch.optim.lr_scheduler.StepLR,
    mask_type='sparsemax',
    verbose=0,
    seed=42
)

tabnet_metrics = train_evaluate_tabnet_model(
    tabnet_model,
    X_train,
    y_train_indexed,
    X_test,
    y_test_indexed,
    model_name="TabNet"
)

all_model_performance.append(tabnet_metrics)


# CNN (using default parameters)
cnn_model = build_keras_cnn_model(
    input_shape=(
        X_train_reshaped_cnn.shape[1],
        X_train_reshaped_cnn.shape[2]
    ),
    num_classes=len(np.unique(y_train_indexed))
)

cnn_metrics = train_evaluate_keras_model(
    cnn_model,
    X_train_reshaped_cnn,
    y_train_indexed,
    X_test_reshaped_cnn,
    y_test_indexed,
    epochs=50,
    batch_size=32,
    model_name="CNN"
)

all_model_performance.append(cnn_metrics)


# LSTM (using default parameters and current data representation)
lstm_model = build_keras_lstm_model(
    input_shape=(
        X_train_reshaped_lstm.shape[1],
        X_train_reshaped_lstm.shape[2]
    ),
    num_classes=len(np.unique(y_train_indexed))
)

lstm_metrics = train_evaluate_keras_model(
    lstm_model,
    X_train_reshaped_lstm,
    y_train_indexed,
    X_test_reshaped_lstm,
    y_test_indexed,
    epochs=50,
    batch_size=32,
    model_name="LSTM"
)

all_model_performance.append(lstm_metrics)


#---Train and Evaluate Ensemble Model--

# Ensemble (VotingClassifier) using Tuned SVM and RandomForestClassifier
estimation = [
    ('svm', tuned_svm_model),
    ('rf', rf_model)
]

voting_clf_subset = VotingClassifier(
    estimators=estimation,
    voting='hard'
)

voting_clf_subset.fit(X_train, y_train)
y_pred_ensemble = voting_clf_subset.predict(X_test)

ensemble_metrics = get_evaluation_metrics(
    y_test,
    y_pred_ensemble,
    "Ensemble (SVM+RF)",
    labels_type='original'
)

all_model_performance.append(ensemble_metrics)


#---4. Comparative Analysis and Visualization--

# Create a DataFrame for model performance
performance_df = pd.DataFrame(all_model_performance)

# Reorder columns for better display
performance_df = performance_df[
    ['Model', 'Accuracy', 'Macro Avg F1', 'Weighted Avg F1']
]

print("\n---Model Performance Summary---")
display(
    performance_df.sort_values(
        by='Accuracy',
        ascending=False
    )
)

# Melt the DataFrame for plotting
performance_melted = performance_df.melt(
    id_vars='Model',
    var_name='Metric',
    value_name='Score'
)

# Create the bar plot
plt.figure(figsize=(16, 8))

sns.barplot(
    x='Model',
    y='Score',
    hue='Metric',
    data=performance_melted
)

plt.title(
    'Model Performance Comparison (Accuracy, Macro Avg F1, Weighted Avg F1)'
)
plt.ylabel('Score')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1.05)
plt.tight_layout()
plt.savefig('model_performance_comparison.png', bbox_inches='tight') # Save the figure
plt.show()


#---Confusion Matrix Visualization for Top Models--

print("\n---Confusion Matrices for Top Models---")

# Identify top models based on Accuracy
top_models_list = performance_df.sort_values(
    by='Accuracy',
    ascending=False
)['Model'].tolist()[:4]

# Generate Confusion Matrices for Top Models
for model_metrics in all_model_performance:
    if model_metrics['Model'] in top_models_list:
        model_name = model_metrics['Model']
        y_true = model_metrics['y_true']
        y_pred = model_metrics['y_pred']

        # Determine labels for confusion matrix based on original vs indexed predictions
        if model_name in ['Tuned MLP', 'TabNet', 'CNN', 'LSTM']:
            plot_labels = sorted(np.unique(y_test_indexed))
        else:
            plot_labels = sorted(np.unique(y_test))

        plot_confusion_matrix(
            y_true,
            y_pred,
            f'Confusion Matrix for {model_name}',
            plot_labels
        )

print("\n--- Finished generating Confusion Matrices for Top Models---")
```
